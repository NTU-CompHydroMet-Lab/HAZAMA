## 定位工作

```python
import pandas as pd
import json
import numpy as np

# 假設你的 DataFrame 叫做 emdat_derived

# 1. 定義一個解析 JSON 的函式，處理 NaN 和格式錯誤
def parse_admin_units_safe(x):
    if pd.isna(x) or x == '':
        return []
    try:
        # 轉換字串為 list of dicts
        return json.loads(x)
    except json.JSONDecodeError:
        return []

# 2. 將 Admin Units 從字串轉為 Python List
emdat_derived['admin_list'] = emdat_derived['Admin Units'].apply(parse_admin_units_safe)

# 3. 使用 explode 將含有多個 adm2 的資料列炸開 (複製資料列)
# 這會讓原本一列有多個行政區的，變成多列，每列對應一個行政區
emdat_exploded = emdat_derived.explode('admin_list').reset_index(drop=True)

# 4. 將 admin_list 內的字典展開為獨立的欄位 (方便後續處理)
# 我們先正規化 admin_list 欄位
admin_details = pd.json_normalize(emdat_exploded['admin_list'])

# 將展開的欄位合併回原本的 DataFrame
emdat_final = pd.concat([emdat_exploded, admin_details], axis=1)

# 檢查一下目前的欄位，應該會有 adm2_code, adm2_name, adm1_name 等出現
print("資料處理完成，總筆數:", len(emdat_final))
emdat_final.head()
```

```python
def get_bbox_from_gee(row):
    """
    根據邏輯查詢 GEE 並回傳 Bounding Box
    邏輯順序: adm2_code -> adm2_name + adm1_name -> adm2_name
    """
    event_id = row['event_id']
    country = row['Country'] # 用來輔助名稱搜尋
    
    # 取得欄位值 (處理 NaN)
    adm2_code = row['adm2_code'] if pd.notna(row['adm2_code']) else None
    adm2_name = row['adm2_name'] if pd.notna(row['adm2_name']) else None
    adm1_name = row['adm1_name'] if pd.notna(row['adm1_name']) else None
    
    # 處理經緯度 (確保是浮點數且非空值)
    try:
        lon = float(row['Longitude'])
        lat = float(row['Latitude'])
        has_coords = not (np.isnan(lon) or np.isnan(lat))
    except (ValueError, TypeError):
        has_coords = False
        lon, lat = None, None
    
    target_feature = None
    match_method = None

    # --- 邏輯 (1): 優先使用 adm2_code ---
    if adm2_code is not None:
        # GAUL 的 ADM2_CODE 通常是整數，確保型態正確
        try:
            code_int = int(adm2_code)
            filtered = gaul.filter(ee.Filter.eq('ADM2_CODE', code_int))
            # 檢查是否有結果
            if filtered.size().getInfo() > 0:
                target_feature = filtered.first()
                match_method = 'ADM2_CODE'
        except:
            pass # 轉換失敗或 API 錯誤就往下走

    # --- 邏輯 (2): 若無 code，使用 adm2_name + adm1_name + Country ---
    if target_feature is None and adm2_name is not None and adm1_name is not None:
        # 為了避免全球同名，我們加入 Country (ADM0) 限制 (如果資料集國家名稱大致吻合)
        # 這裡示範標準邏輯：ADM2 + ADM1
        filtered = gaul.filter(ee.Filter.and_(
            ee.Filter.eq('ADM0_NAME', country),
            ee.Filter.eq('ADM2_NAME', adm2_name),
            ee.Filter.eq('ADM1_NAME', adm1_name)
        ))
        
        # 如果找不到，嘗試放寬條件，加入 Country 限制但只看 ADM2+ADM1
        if filtered.size().getInfo() > 0:
            target_feature = filtered.first()
            match_method = 'ADM0&1&2_NAME'

    # --- 邏輯 (3): 若無上述，僅使用 adm2_name + Country ---
    if target_feature is None and adm2_name is not None:
        filtered = gaul.filter(ee.Filter.and_(
            ee.Filter.eq('ADM0_NAME', country),
            ee.Filter.eq('ADM2_NAME', adm2_name)
        ))
        
        # 如果找到多個，嘗試用 Country 過濾 (需注意資料集國家名稱拼寫差異)
        # 這裡簡化邏輯，直接取第一個匹配的結果
        if filtered.size().getInfo() > 0:
            target_feature = filtered.first()
            match_method = 'ADM0&2_NAME'

    # --- 邏輯 (4): 若無法透過名稱定位，且經緯度有值，則用座標反查 ---
    if target_feature is None and has_coords:
        # 建立幾何點
        point = ee.Geometry.Point([lon, lat])
        # 使用 filterBounds 找出包含該點的行政區
        filtered = gaul.filterBounds(point)
        
        if filtered.size().getInfo() > 0:
            target_feature = filtered.first()
            match_method = 'Coordinates_Lookup'
    
    # --- 邏輯 (5): 若都不符合，回傳錯誤訊息 ---
    if target_feature is None:
        # print(f"event_id: {event_id} cannot be located")
        return {
            'bbox': None,
            'match_method': 'cannot_be_located' # 這裡會顯示在表格中
        }

    # --- 取得 Bounding Box 座標 ---
    try:
        # 取得幾何範圍
        geom = target_feature.geometry()
        # bounds() 回傳的是一個矩形 Polygon
        # bounds 格式通常是 [[minx, miny], [maxx, miny], [maxx, maxy], [minx, maxy], [minx, miny]]
        bounds = geom.bounds().coordinates().get(0).getInfo()
        
        return {
            'bbox': bounds[0:4],
            'match_method': match_method
        }
    except Exception as e:
        # 若發生 GEE 內部幾何錯誤，也標記失敗
        return {
            'bbox': None,
            'match_method': f'error_{str(e)}'
        }
```

```python
from tqdm import tqdm
tqdm.pandas()  # 啟動 pandas 的進度條功能

# 為了測試，我們先取前 10 筆 (正式執行時請拿掉 .head(10))
df_to_process = emdat_final.iloc[0:10].copy() 
# df_to_process = emdat_final.copy()

print("開始向 GEE 查詢 Bounding Box (這會花一點時間)...")

# 使用 apply 執行查詢
# 結果會存成一個 dict，稍後再拆開
df_to_process['bbox_result'] = df_to_process.progress_apply(get_bbox_from_gee, axis=1)

# 將結果拆分成獨立欄位
bbox_df = pd.json_normalize(df_to_process['bbox_result'])
final_df = pd.concat([df_to_process.reset_index(drop=True), bbox_df], axis=1)

# 顯示結果
# 檢查哪些地點找不到 (這裡就會顯示 'cannot_be_located')
missing_locations = final_df[final_df['match_method'] == 'cannot_be_located']
print(f"共有 {len(missing_locations)} 筆資料無法定位")

final_df[['event_id', 'Country', 'adm2_name','Longitude', 'Latitude', 'bbox', 'match_method', 'start_date', 'end_date']]
```
